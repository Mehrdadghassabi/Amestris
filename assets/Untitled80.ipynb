```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_excel('Evaluation_Results_Qualitative.xlsx')
```

```python
total = 0
tie = 0
ames = 0
gem = 0
for index, row in df.iterrows():
    total += 1
    if row['Evaluation'].startswith("Tie"):
       tie += 1
    elif row['Evaluation'].startswith("LORA"):
       ames += 1
    elif row['Evaluation'].startswith("Baseline"):
       gem += 1
    else:
       print("None")
print(total)
print(tie)
print(ames)
print(gem)
```

```python
import matplotlib.pyplot as plt

# Your variables
labels = ['amestris-1b-dpo Win', 'Tie', 'gemma3-1b Win']
sizes = [100*ames/total, 100*tie/total, 100*gem/total]
colors = ['#2ca02c', '#7f7f7f', '#d62728'] # Green, Gray, Red
explode = (0.05, 0, 0)

fig, ax = plt.subplots(figsize=(6, 6))

# Pass labels to wedges, but don't show external text layout
wedges, texts, autotexts = ax.pie(
    sizes,
    explode=explode,
    labels=None,  # Hides the ugly external text/pointers
    colors=colors,
    autopct='%1.1f%%',
    startangle=140,
    pctdistance=0.8,
    textprops=dict(color="black"),
    wedgeprops=dict(width=0.4, edgecolor='w')
)

# --- CENTER LEGEND WITH COLORS ---
ax.legend(
    wedges,
    labels,
    title="",
    title_fontproperties={'weight': 'bold', 'size': 11},
    loc="center",         # Moves it to (0,0)
    frameon=False,        # Removes the background box/border
    fontsize=10,
    handletextpad=0.5,    # Space between color box and text
    labelspacing=0.6      # Vertical space between rows
)
# ---------------------------------

# Style the percentage text inside the rings to be white for better visibility
plt.setp(autotexts, size=10, weight="bold", color="white")

ax.set_title("", fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('fig3.jpg', dpi=300)
```
